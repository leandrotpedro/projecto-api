# Documentação da CategoryController

A controller gerencia operações de consulta, criação, atualização e exclusão de categorias. Os caminhos das rotas dependem da configuração em `routes/api.php`.

## 1. Listar categorias

**Método:** `index`

Retorna todas as categorias, incluindo a contagem de produtos associados a cada uma.

**Resposta de sucesso:** `200 OK`

O corpo contém uma lista de categorias com a contagem de produtos.

## 2. Criar categoria

**Método:** `store`

Cria uma categoria com os dados enviados no corpo da requisição.

**Validações:**
- `name`: obrigatório, texto, máximo de 255 caracteres e único na tabela `categories`.
- `description`: opcional e deve ser texto quando informado.

**Resposta de sucesso:** `201 Created`

Retorna uma mensagem de sucesso e os dados da categoria criada.

**Erro de validação:** `422 Unprocessable Entity`

## 3. Consultar categoria

**Método:** `show`

Busca uma categoria pelo identificador e inclui seus produtos associados.

**Resposta de sucesso:** `200 OK`

Retorna os dados da categoria e seus produtos.

**Categoria inexistente:** `404 Not Found`

Retorna a mensagem `Categoria não encontrada.`

## 4. Atualizar categoria

**Método:** `update`

Atualiza os dados da categoria identificada. Os campos podem ser omitidos; se `name` for enviado, ele deve ser único, desconsiderando a própria categoria.

**Validações:**
- `name`: opcional; quando informado, é obrigatório, deve ser texto, ter no máximo 255 caracteres e ser único.
- `description`: opcional e pode ser nulo; quando informado, deve ser texto.

**Resposta de sucesso:** `200 OK`

Retorna uma mensagem de sucesso e os dados atualizados.

**Categoria inexistente:** `404 Not Found`

Retorna a mensagem `Categoria não encontrada.`

**Erro de validação:** `422 Unprocessable Entity`

## 5. Eliminar categoria

**Método:** `destroy`

Exclui a categoria identificada somente quando ela não possui produtos associados.

**Resposta de sucesso:** `200 OK`

Retorna uma mensagem de confirmação.

**Categoria inexistente:** `404 Not Found`

Retorna a mensagem `Categoria não encontrada.`

**Categoria com produtos associados:** `422 Unprocessable Entity`

A exclusão é impedida e a resposta informa que há produtos associados.
